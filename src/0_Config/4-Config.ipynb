```python
def create_catalog_and_schemas():
    spark.sql("CREATE CATALOG IF NOT EXISTS case_databricks")
    spark.sql("CREATE SCHEMA IF NOT EXISTS case_databricks.landing")
    spark.sql("CREATE SCHEMA IF NOT EXISTS case_databricks.bronze")
    spark.sql("CREATE SCHEMA IF NOT EXISTS case_databricks.silver")
    spark.sql("CREATE SCHEMA IF NOT EXISTS case_databricks.gold")
    spark.sql("CREATE SCHEMA IF NOT EXISTS case_databricks.controller")
    spark.sql("CREATE SCHEMA IF NOT EXISTS case_databricks.sandbox")
    spark.sql("CREATE SCHEMA IF NOT EXISTS case_databricks.platinum")

create_catalog_and_schemas()
```

```python
def create_landing_managed_volume():
    """
    Cria o Managed Volume 'landing_volume' no schema 'case_databricks.landing' se não existir.
    """
    spark.sql("""
    CREATE MANAGED VOLUME IF NOT EXISTS case_databricks.landing.landing_volume
    """)
```
